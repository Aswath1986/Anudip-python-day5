The format has only a few fields:

| Field | Meaning |
|---|---|
| `type` | `code` (player types Python) or `predict` (player says what the code prints) |
| `topic` | One or two words, such as `strings` |
| `task` | What the player must do |
| `setup` | Optional. One line of Python that runs first, to create names the player can use |
| `answer` | For `code` questions: one correct line of Python. You write this |
| `hidden` | The same answer, scrambled so that nobody reads it by accident on GitHub. The game makes this for you in Part 5 |
| `must_use` | Optional. A word or operator the player's answer must contain |
| `code` | For `predict` questions: the code to show. It must be the **last** field |

## Part 5: Write your own question

Change the values below. Use something from last week that **you** found tricky, because other people will find it tricky too.

Rules for a good question:

- The task says clearly what to do.
- The answer is one line of Python.
- Test it yourself before you send it.

````python
MY_NAME = input("Your first name, no spaces: ").strip().lower()

my_question = """type: code
topic: strings
task: Count the characters in the name called word
setup: word = "python"
answer: len(word)
"""

my_file = f"questions/students/{MY_NAME}.txt"
with open(my_file, "w") as file:
    file.write(my_question)
print("Saved:", my_file)
````

## Part 6: Send your question through GitHub

Open a terminal in the repo folder and run one line at a time. Replace `<YOURNAME>`.

```cmd
git checkout -b question-<YOURNAME>
git status
git add SyntaxQuiz/questions/students
git commit -m "Add quiz question by <YOURNAME>"
git push -u origin question-<YOURNAME>
```

Then open your fork on GitHub and click **Compare & pull request**, then **Create pull request**.

The facilitator reads each question before merging it. In industry this is called a **code review**: nothing joins the shared code until a second person has read it.
